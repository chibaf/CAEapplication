```python
#%matplotlib widget
import numpy as np
from matplotlib import pyplot as plt
import ipywidgets as widgets

plt.rcParams['font.family']      = 'Times New Roman'
plt.rcParams['mathtext.fontset'] = 'stix'
plt.rcParams['text.usetex']      = True
plt.rcParams['font.size']        = 12
plt.rcParams['axes.labelsize']   = 'xx-large'
```

```python
def Erms (X, Y):
    return np.sqrt(np.mean((X - Y) ** 2))
```

```python
# 
#   Definition of data set
#
xtrain = np.array([ 0.02, 0.22, 0.44,  0.67,  0.90, 0.16, 0.34, 0.50,  0.73, 1.00 ])
ytrain = np.array([-0.02, 0.97, 0.25, -0.90, -0.53, 0.94, 0.85, 0.12, -0.93, 0.08 ])
X_t = np.mat(xtrain).T
Y_t = np.mat(ytrain).T
# %%
x_g = np.linspace(0, 1, 101)
y   = np.sin(2.0 * np.pi * x_g)
X_g = np.mat(x_g).T
#
nt  = len(X_t)
ng  = len(X_g)
```

```python
def Draw(deg):
    Xd = np.mat(np.empty((nt, deg+1)))
    Xg = np.mat(np.empty((ng, deg+1)))
    for i in range(deg+1):
        Xd[:, i] = np.power(X_t, i)
        Xg[:, i] = np.power(X_g, i)    
    # %%
    fig, ax = plt.subplots(dpi=100)
    ax.scatter(xtrain, ytrain, marker='x', color='k', linewidth=2, label='training Data', zorder=3)
    ax.plot(x_g, y, linestyle='dotted', label=r'$y\,=\,\sin\,(2\pi\,x)$')
    #
    lmd = [0, 1e-4, 1e-2]
    ls  = ['solid', 'dashdot','dashed']
    for i in range(3):
        lambd = lmd[i]
        w = (Xd.T * Xd + lambd * np.eye(deg+1)).I * Xd.T * Y_t
        E = Erms(np.array(Xd * w), np.array(Y_t))        
        ax.plot(x_g, Xg * w, label=r'$\lambda=%.2e$   $E_{\rm rms}=%.3f$' % (lambd, E), ls=ls[i])
    ax.set_xlabel('$x$')
    ax.set_ylabel('$y$', rotation='horizontal')
    ax.set_ylim([-1.5, 1.5])
    ax.grid(linestyle='--', alpha=0.5)
    ax.legend(loc='upper right', fontsize='small')
    ax.set_title('Overfitting and Regularization  ( degree=$%d$ )' % deg)
    plt.show()
```

```python
def main():
    s1  = widgets.IntSlider(value=5,   min=1,  max=11,  step=1,  description='degree:')
    out = widgets.interactive_output(Draw, {'deg': s1})
    ui  = widgets.HBox([s1, ])
    #
    display(ui, out)

if __name__ == '__main__':
    main()
```
